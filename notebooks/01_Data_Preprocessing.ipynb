```python
id_cols = [
    "Daily_Rate",
    "Total_Rental_Amount",
    "Total_Rental_Amount_Original",
    "Security_Deposit",
    "Delivery_Charge",
    "Maintenance_Cost"
]

for col in id_cols:
    print(f"{col}: {(df[col] < 0).sum()} negative values")
```

```python
outlier_columns = [
    "Rental_Duration_Days",
    "Daily_Rate",
    "Total_Rental_Amount",
    "Security_Deposit",
    "Delivery_Charge",
    "Maintenance_Cost",
    "Customer_Rating",
    "Equipment_Age_Years"
]

for col in outlier_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ][col].notna().sum()

    print(
        f"{col}: {outliers} outliers "
        f"(lower={lower_bound:.2f}, upper={upper_bound:.2f})"
    )
```
