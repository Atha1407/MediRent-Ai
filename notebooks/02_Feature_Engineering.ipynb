```python
monthly_demand = (
    df.groupby(
        [
            df["Rental_Start_Date"].dt.to_period("M"),
            "Equipment"
        ]
    )
    .size()
    .reset_index(name="Rental_Count")
)

monthly_demand.columns = [
    "Rental_Month_Period",
    "Equipment",
    "Rental_Count"
]

print("Shape:", monthly_demand.shape)
monthly_demand.head(20)
```

```python
# Create a proper datetime column for the monthly demand table

monthly_demand["Month"] = monthly_demand["Rental_Month_Period"].dt.to_timestamp()

# Sort chronologically within each equipment type
monthly_demand = monthly_demand.sort_values(
    ["Equipment", "Month"]
).reset_index(drop=True)

# Previous month's demand for the same equipment
monthly_demand["Historical_Demand_Lag_1"] = (
    monthly_demand.groupby("Equipment")["Rental_Count"].shift(1)
)

print(monthly_demand.head(20))
```

```python
monthly_demand["Historical_Demand_Rolling_3M"] = (
    monthly_demand
    .groupby("Equipment")["Rental_Count"]
    .transform(
        lambda x: x.shift(1).rolling(window=3, min_periods=1).mean()
    )
)

print(monthly_demand.head(10))
```

```python
monthly_demand_final = (
    df.groupby(
        [
            df["Rental_Start_Date"].dt.to_period("M"),
            "Equipment"
        ]
    )
    .size()
    .reset_index(name="Monthly_Demand")
)

monthly_demand_final.columns = [
    "Month",
    "Equipment",
    "Monthly_Demand"
]

monthly_demand_final = monthly_demand_final.sort_values(
    ["Month", "Equipment"]
).reset_index(drop=True)

print("Monthly demand dataset shape:", monthly_demand_final.shape)

print("\nFirst 10 rows:")
print(monthly_demand_final.head(10))

print("\nLast 10 rows:")
print(monthly_demand_final.tail(10))
```

```python
all_months = pd.period_range(
    start=df["Rental_Start_Date"].min().to_period("M"),
    end=df["Rental_Start_Date"].max().to_period("M"),
    freq="M"
)

all_equipment = sorted(df["Equipment"].dropna().unique())

complete_grid = pd.MultiIndex.from_product(
    [all_months, all_equipment],
    names=["Month", "Equipment"]
).to_frame(index=False)

missing_combinations = complete_grid.merge(
    monthly_demand_final,
    on=["Month", "Equipment"],
    how="left"
)

missing_combinations = missing_combinations[
    missing_combinations["Monthly_Demand"].isna()
]

print("Expected combinations:", len(complete_grid))
print("Observed combinations:", len(monthly_demand_final))
print("Missing combinations:", len(missing_combinations))

print("\nMissing equipment-month combinations:")
print(missing_combinations.to_string(index=False))
```

```python
monthly_demand_final = (
    complete_grid.merge(
        monthly_demand_final,
        on=["Month", "Equipment"],
        how="left"
    )
)

monthly_demand_final["Monthly_Demand"] = (
    monthly_demand_final["Monthly_Demand"]
    .fillna(0)
    .astype(int)
)

monthly_demand_final = monthly_demand_final.sort_values(
    ["Month", "Equipment"]
).reset_index(drop=True)

print("Final monthly demand dataset shape:", monthly_demand_final.shape)

print("\nDemand statistics:")
print(monthly_demand_final["Monthly_Demand"].describe())

print("\nNumber of zero-demand equipment-months:")
print((monthly_demand_final["Monthly_Demand"] == 0).sum())

print("\nMissing target values:")
print(monthly_demand_final["Monthly_Demand"].isna().sum())
```

```python
monthly_demand_final = monthly_demand_final.sort_values(
    ["Equipment", "Month"]
).reset_index(drop=True)

monthly_demand_final["Historical_Demand_Lag_1"] = (
    monthly_demand_final
    .groupby("Equipment")["Monthly_Demand"]
    .shift(1)
)

monthly_demand_final["Historical_Demand_Lag_2"] = (
    monthly_demand_final
    .groupby("Equipment")["Monthly_Demand"]
    .shift(2)
)

monthly_demand_final["Historical_Demand_Rolling_3M"] = (
    monthly_demand_final
    .groupby("Equipment")["Monthly_Demand"]
    .transform(
        lambda x: x.shift(1).rolling(
            window=3,
            min_periods=1
        ).mean()
    )
)

monthly_demand_final["Historical_Demand_Rolling_6M"] = (
    monthly_demand_final
    .groupby("Equipment")["Monthly_Demand"]
    .transform(
        lambda x: x.shift(1).rolling(
            window=6,
            min_periods=1
        ).mean()
    )
)

monthly_demand_final["Historical_Demand_Trend_2M"] = (
    monthly_demand_final["Historical_Demand_Lag_1"]
    - monthly_demand_final["Historical_Demand_Lag_2"]
)

print("Shape:", monthly_demand_final.shape)

print("\nColumns:")
print(monthly_demand_final.columns.tolist())

print("\nMissing values:")
print(monthly_demand_final.isna().sum())
```

```python
# Create month name from the month number
monthly_demand_final["Month_Name"] = (
    monthly_demand_final["Month_Number"]
    .map(lambda x: pd.Timestamp(2024, x, 1).month_name())
)

print("Shape:", monthly_demand_final.shape)

print("\nColumns:")
print(monthly_demand_final.columns.tolist())

print("\nSample:")
print(monthly_demand_final.head(10).to_string(index=False))
```

```python
# Create Equipment → Equipment_Category mapping
equipment_category_map = (
    df[["Equipment", "Equipment_Category"]]
    .drop_duplicates()
)

# Verify each equipment has only one category
category_check = (
    equipment_category_map
    .groupby("Equipment")["Equipment_Category"]
    .nunique()
)

print("Equipment types with multiple categories:",
      (category_check > 1).sum())

# Add category to monthly demand dataset
monthly_demand_final = monthly_demand_final.merge(
    equipment_category_map,
    on="Equipment",
    how="left"
)

print("\nShape:", monthly_demand_final.shape)
print("\nMissing Equipment_Category:",
      monthly_demand_final["Equipment_Category"].isna().sum())

print("\nSample:")
print(
    monthly_demand_final[
        ["Month", "Equipment", "Equipment_Category", "Monthly_Demand"]
    ].head(10).to_string(index=False)
)
```

```python
demand_model_features = [
    "Equipment",
    "Equipment_Category",
    "Year",
    "Month_Number",
    "Quarter",
    "Historical_Demand_Lag_1",
    "Historical_Demand_Lag_2",
    "Historical_Demand_Rolling_3M",
    "Historical_Demand_Rolling_6M",
    "Historical_Demand_Trend_2M"
]

demand_model_df = monthly_demand_final[
    demand_model_features + ["Monthly_Demand"]
].copy()

print("Demand model dataset shape:", demand_model_df.shape)

print("\nColumns:")
print(demand_model_df.columns.tolist())

print("\nMissing values:")
print(demand_model_df.isna().sum())

print("\nDuplicate rows:")
print(demand_model_df.duplicated().sum())
```
